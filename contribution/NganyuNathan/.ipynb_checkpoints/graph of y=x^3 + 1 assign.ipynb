```python
import numpy as np
```

```python
x=np.array([1,2,3,4,5])
y=np.array([1,3,4,5,5])
```

```python
def y_cube(x):
    return x**3 + 1
```

```python
x=np.array([1,2,3,4,5])
y_cube(x)
y=y_cube(x)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(x,y, color='navy', linewidth=1.5)
plt.title("graph of y against x")
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```
