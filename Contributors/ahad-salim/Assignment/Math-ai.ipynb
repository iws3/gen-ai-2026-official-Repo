```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([1,2,3,4,5])
def func(x):
    return (x**3)+1
```

```python
func(x)
```

```python
y = func(x)
```

```python
plt.plot(x, y, color='navy', linewidth=2)
plt.title("Graph of y against x")
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```
