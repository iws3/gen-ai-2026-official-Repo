```python
# CREATE FUNCTION
def y_squared(x):
    return x**2
```

```python
y_squared(2)
```

```python
x=np.array([1, 2, 3,4,5])
y_squared(x)
```

```python
y=y_squared(x)
```

```python
# PLOT A GRAPH OF Y AGAINST X
```

```python
# WE ARE GOING TO BE USING MATPLOTLIB
import matplotlib.pyplot as plt
```

```python
plt.plot(x, y, color='navy', linewidth=2)
plt.title("Graph of y against x")
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```
