# 1.2 What Is a Vector?
An ordered list of numbers. It has both magnitude (length) and direction. Think of it as an
arrow pointing somewhere in space.


```python
# creating vectors in numpy
v=np.array([1,2, 3.7])
v
u=np.array([1,2,3,4])
```

```python
#  CHECK THE SHAPE OF YOUR VECTOR

print(v.shape)
print(u.shape)
print(v.ndim)
print(v.dtype)
```

# Indexing a Vector
Indexing works identically to Python lists — zero-based:

```python
v = np.array([10, 20, 30, 40, 50])

print(v[0])
print(v[-1])
print(v[:4])
```

## Row vs Column Vectors
A row vector is shape (n,) or (1, n). A column vector is shape (n, 1). The distinction becomes critical in
matrix multiplication:


```python
v=np.array([1,2,5])
print(v.shape)
```

```python
row=v.reshape(1, 3)
col=v.reshape(3, 1)
print(row.shape)
print(col.shape)
```

# Scalar multiplication

```python
v = np.array([2, 4, 6])
scalar=2
result=scalar * v;
print(result)
```

```python
import matplotlib.pyplot as plt

a=np.array([3,1])
b=np.array([1,3])
c=a+b
print(c)

fig , ax=plt.subplots(figsize=(6, 4))

ax.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1,
 color='navy', label='a = [3,1]')
ax.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1,
 color='gold', label='b = [1,3]')
ax.quiver(0, 0, c[0], c[1], angles='xy', scale_units='xy', scale=1,
 color='red', label='a+b = [4,4]', linestyle='dashed')
ax.set_xlim(-1, 6)
ax.set_ylim(-1, 6)
```
